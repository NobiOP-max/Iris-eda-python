```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from IPython.display import display


```

```python
from ipywidgets import FileUpload
uploader = FileUpload(accept='.csv', multiple=False)
display(uploader)
```

```python
import io
import pandas as pd
if len(uploader.value) == 0:
    print("Please upload a CSV file using the Upload button above.")
else:
    uploaded_file = uploader.value[0]

    content = uploaded_file.content

    df = pd.read_csv(io.BytesIO(content))

    print("Dataset loaded successfully!")
    display(df.head())
```

```python
print("Shape:", df.shape)
print("\nInfo:")
df.info()
print("\nStatistical Summary:")
df.describe()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df.drop(columns=['Species']))
plt.title("Outlier Detection using Boxplot")
plt.show()
```

```python
le = LabelEncoder()
df['Species_Encoded'] = le.fit_transform(df['Species'])
df[['Species', 'Species_Encoded']].head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

num_cols = [
    'SepalLengthCm',
    'SepalWidthCm',
    'PetalLengthCm',
    'PetalWidthCm'
]

df[num_cols] = scaler.fit_transform(df[num_cols])

df.head()

```

```python
df[num_cols].hist(figsize=(10,8))
plt.suptitle("Feature Distribution After Normalization")
plt.show()
```

```python
sns.scatterplot(
 x=df['PetalLengthCm'],
 y=df['PetalWidthCm'],
 hue=df['Species']
)
plt.title("Petal Length vs Petal Width")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(
 df.drop(columns=['Species']).corr(),
 annot=True,
 cmap='coolwarm'
)
plt.title("Correlation Heatmap")
plt.show()
```
